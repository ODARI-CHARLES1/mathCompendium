Plotting the data(scatter) helps to determine the relationship between the data

```python
def scatter_plot(x,y):
    plt.scatter(x,y)
    plt.xlabel("x")
    plt.ylabel("y")
    plt.title("Scatter plot of the data")
    plt.show()
```

# Imports


```python
import numpy as np
import scipy as scp
import scienceplots
import sympy as sp
import matplotlib.pyplot as plt

plt.rcParams["text.usetex"] = False
plt.style.use(["science", "no-latex"])
```

```python
from sympy.abc import x
```

plot of a rational function

$f=\frac{x-1}{x+4}$

```python
#define the function

f=(x-1)/(x+4)
f
```

```python
#lambdify the function 
y_func=sp.lambdify(x,f,["numpy","scipy"])
x_values=np.linspace(-5,0,1000)
```

```python
#plot the function
fig,ax=plt.subplots(1,figsize=(10,4))
ax.plot(x_values,y_func(x_values))
ax.set_title("Plot of (x-1)(x+4)")
ax.set_xlabel("x")
ax.set_ylabel("y")
plt.grid(True,linestyle="--",alpha=0.4)
```

```python
#save the plot
fig.savefig("plot.png", dpi=100, bbox_inches="tight")
```

```python
import matplotlib.pyplot  as plt
import pandas as pd
import numpy as np
import scienceplots
import warnings
import os

warnings.filterwarnings("ignore")
plt.style.use(["science","no-latex"])
```

```python
data=pd.read_csv("data.csv")
data.head()
```

```python
fig,ax=plt.subplots(figsize=(10,4))
plt.scatter(data.x,data.y,marker="x",label="Non-Linear model")
plt.scatter(np.log1p(data.x),np.log1p(data.y),label="Logarithmic Linear")
plt.grid(True,linestyle='--',alpha=0.3)
plt.legend()
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

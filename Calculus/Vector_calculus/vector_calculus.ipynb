```python
from mpl_toolkits.mplot3d import Axes3D
import matplotlib.pyplot as plt
import numpy as np

# Define points A and B
A = np.array([0, 0, 0])
B = np.array([3, 2, 5])

# Parametric line r(t) = A + t(B-A)
t = np.linspace(0, 1, 100)
line = A[:, None] + np.outer(B - A, t)

x = line[0]
y = line[1]
z = line[2]

# Plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.plot(x, y, z)
ax.scatter(A[0], A[1], A[2])
ax.scatter(B[0], B[1], B[2])

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Plane coefficients
a, b, c, d =2, 3, 4, -12   # example: x + 2y + 3z - 6 = 0

# Create grid for x and y
x = np.linspace(-50, 5, 50)
y = np.linspace(-50, 5, 50)
X, Y = np.meshgrid(x, y)

# Solve for Z
Z = (-a*X - b*Y - d) / c

# Plot
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
ax.plot_surface(X, Y, Z, alpha=1)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

plt.show()
```

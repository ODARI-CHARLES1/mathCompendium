## Equation of a plane 

$ax+bx+cx+d=0$ 

where:

$<a,b,c>$ 

is the unit vector of the normal vector



```python
import numpy as np
import matplotlib.pyplot as plt

# Plane coefficients
a, b, c, d =2, 3, 4, -12   # example: x + 2y + 3z - 6 = 0

# Create grid for x and y
x = np.linspace(-50, 5, 50)
y = np.linspace(-50, 5, 50)
X, Y = np.meshgrid(x, y)

# Solve for Z
Z = (-a*X - b*Y - d) / c

# Plot
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
ax.plot_surface(X, Y, Z, alpha=1)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

plt.show()
```

## Quadric Surfaces

A quadric surfaceis the graph of a second-degree equation in three variables x, y, and z.
The most general such equation is:

$ Ax^2+By^2+Cz^2+Dxy+Eyz+Fxz+Gx+Hy+Iz+J=0 $

where , , are constants, but by translation and rotation it can be brought into one
of the two standard forms

- $Ax^2+By^2+Cz^2+J=0$

- $Ax^2+By^2+Cz+J=0$

Examples:

- Sphere
- Ellipsoid
- Hyperboloid
- Paraboloid
- Cone

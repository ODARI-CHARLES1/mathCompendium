# Integral of a vector Functions

The definite integral of a continuous vector function can be defined in much the same
way as for real-valued functions except that the integral is a vector.


```python
import numpy as np
import scipy as scp
import sympy as sp
import matplotlib.pyplot as plt
```

```python
from sympy.abc import t,k,i,j
```

## Determine the integral 

$r(t)=(1+t^3)i+te^{-t}j+te^{-t}j+sin(2t)k$

```python
r=(1+t**3)*i+t*sp.exp(-t)*j+sp.sin(2*t)*k
r
```

```python
sp.integrate(r,t)
```

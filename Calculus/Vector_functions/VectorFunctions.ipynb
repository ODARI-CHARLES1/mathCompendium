# Vector Functions
A **vector-valued function**, or **vector function**, is simply a function whose domain is
a set of real numbers and whose range is a set of vectors. 

# Imports 

```python
import numpy as np
import scipy 
import sympy as sp
import matplotlib.pyplot as plt
import scienceplots
plt.style.use(['science', 'no-latex'])
plt.style.available
```

# Determine Limits of  a Function



```python
from sympy.abc import t,i,j,k
```

```python
#Define the function
r=(1+t**3)*i+t*sp.exp(-t)*j+(sp.sin(t)/t)*k

#determine the limit
sp.limit(r,t,0)
```

### Visualise the vector function


```python

x_expr = 1 + t**3
y_expr = t * sp.exp(-t)
z_expr = sp.sin(t) / t

# 2. Compute the derivative (velocity/tangent vector)
dx = sp.diff(x_expr, t)
dy = sp.diff(y_expr, t)
dz = sp.diff(z_expr, t)

# 3. Convert symbolic expressions to numerical functions
x_func = sp.lambdify(t, x_expr, 'numpy')
y_func = sp.lambdify(t, y_expr, 'numpy')
z_func = sp.lambdify(t, z_expr, 'numpy')

dx_func = sp.lambdify(t, dx, 'numpy')
dy_func = sp.lambdify(t, dy, 'numpy')
dz_func = sp.lambdify(t, dz, 'numpy')

# 4. Generate points for t (avoiding exactly t=0 to prevent division by zero)
t_vals = np.linspace(0.01, 5.0, 50)

# Calculate positions (r)
X = x_func(t_vals)
Y = y_func(t_vals)
Z = z_func(t_vals)

# Calculate vector directions (tangents)
U = dx_func(t_vals)
V = dy_func(t_vals)
W = dz_func(t_vals)
```

```python
# 5. Plotting in 3D
fig = plt.figure(figsize=(10,7))
ax = fig.add_subplot(111, projection='3d')

# Plot the path of the curve
ax.plot(X, Y, Z, label='r(t) path', color='blue', linewidth=2)

# Plot the tangent vectors along the path
# Set length parameter to scale arrows visually
ax.quiver(X, Y, Z, U, V, W, length=0.2, color='red', normalize=True, label='Tangent Vectors')

ax.set_title("3D Vector Field of $\\vec{r}(t)$")
ax.set_xlabel('X axis')
ax.set_ylabel('Y axis')
ax.set_zlabel('Z axis')
ax.legend()

plt.show()
```

# Plot the intersectin of a cylinder and a plane

$Cylinder: x^2+y^2=1$

$Plane: x+z=2$

```python
z=np.linspace(-5,5,1000)
theta=np.linspace(0,2*np.pi,1000)

z_grid,theta_grid=np.meshgrid(z,theta)
X,Y=np.meshgrid(z,z)

#cylinder points
r=1
x=r*np.cos(theta_grid)
y=r*np.sin(theta_grid)
z=z_grid

#plane 
Z=-Y+2

fig=plt.figure(figsize=(20,8))
ax=fig.add_subplot(111,projection="3d")
ax.plot_surface(x,y,z,label="cylinder")
ax.plot_surface(X,Y,Z,label="Plane")
ax.set_title("intersection of cylinder and plane intersection")
ax.legend()
plt.show()
```

# Derivatives of Vector Functions

Derivative of a vector functions is defined the same way as for real values

$\frac{dr}{dt} =r'(t)=\lim_{h \to 0}\frac{r(t-h)-r(t)}{h}$

```python
import numpy as np
import scipy 
import sympy as sp
import matplotlib.pyplot as plt
import scienceplots
plt.style.use(['science', 'no-latex'])
```

```python
from sympy.abc import t,i,j,k
```

```python
r=(1+t**3)*i+t*sp.exp(-t)*j+sp.sin(2*t)*k
r
```

```python
#find the derivative
dr_dt=sp.diff(r,t)
dr_dt
```

```python
#find the value of the tanget at t=0
dr_dt.subs(t,0)
```

### Determine the tanget vector function

let vector functions  =$v(t)$

$v(t)=\frac{r'(0)}{|r'(0)}$

```python
num=dr_dt.subs(t,0)
num_matrix=sp.Matrix([0,1,2])
den=num_matrix.norm()
```

```python
#vector function
v=num/den
v
```

# SYSTEM OF LINEAR EQUATIONS

$3x+4y=7$

$4x+5y=9$

```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
```

```python
from sympy.abc import x,y
```

```python
eq1=sp.Eq(3*x+4*y,7)
eq2=sp.Eq(4*x+5*y,9)
```

```python
eq1
```

```python
eq2
```

```python
#solving the system of linear equations
sp.solve((eq1,eq2),(x,y))
```

```python
#USING NUMPY TO SOLVE
#AX=b
A=np.array([[3,4],[4,5]])
b=np.array([7,9])
```

```python
np.linalg.solve(A,b)
```

```python
np.linalg.inv(A)
```

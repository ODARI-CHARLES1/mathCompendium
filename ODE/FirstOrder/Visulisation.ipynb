```python
import matplotlib.pyplot as plt
import numpy as np
import scipy as scp
from matplotlib import colormaps
from sympy.abc import t,y
from sympy import dsolve,cos,Eq
from scipy.integrate import solve_ivp
```

$y'=cos(t)cos(y)$

```python
_=np.linspace(-2*np.pi,2*np.pi,20)
T,Y=np.meshgrid(_,_)
Dy=np.cos(T)*np.cos(Y)
plt.quiver(T,Y,Dy)
plt.colorbar()
plt.title("Quiver plot of $y'=cos(t)cos(y)$")
plt.show()
```

```python
#3d plot of y'
_=np.linspace(-2*np.pi,2*np.pi,2000)
T,Y=np.meshgrid(_,_)
Dy=np.cos(T)*np.cos(Y)
fig=plt.figure(figsize=(9,6))
ax=fig.add_subplot(projection="3d")
ax.plot_surface(T,Y,Dy,cmap="CMRmap_r")
plt.show()
```

```python
def dydt(t,y):
    return np.cos(t)*np.cos(y)
```

```python
times=np.linspace(0,10,100)
y0=np.linspace(0,1,10)
sol=solve_ivp(dydt,[0,10],y0,t_eval=times)
```

```python
sol.t
```

```python
sol.y
```

```python
plt.figure(figsize=(5,3))
for i in range(len(y0)):
    plt.plot(sol.t,sol.y[i])
plt.grid(True,axis='y',linestyle="--",alpha=0.4)
plt.show()
```
